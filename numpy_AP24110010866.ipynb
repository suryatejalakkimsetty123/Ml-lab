1. Why is NumPy faster than Python lists?#AP24110010866

NumPy stores data in one continuous block of memory using the same data type, and runs operations using fast, pre-compiled C code. Python lists store items separately with more overhead, and loop through them one by one using the slower Python interpreter.

2. Explain broadcasting with an example of your own.#AP24110010866

Broadcasting lets NumPy apply an operation to a whole array without writing a loop. For example, image + 50 adds 50 to every single pixel in the image at once, even though 50 is just one number and the image is a big 3D array.

The image is 512 pixels tall, 512 pixels wide, and has 3 color channels (Red, Green, Blue) at every pixel.#AP24110010866

An image can have millions of numbers, so printing them all is unreadable and useless. It's better to check .shape or use plt.imshow() to actually see the image.

5. Explain RGB channels.#AP24110010866

RGB channels are three separate grids of values — Red, Green, and Blue — that represent color intensity at each pixel. Combining all three in different amounts creates every color in the image.#AP24110010866

```python
import numpy as np  # Import NumPy for vector math
#AP24110010866
# Two similar sentences (same topic: NumPy/data science)#AP24110010866
sentence_A = "NumPy arrays make data science calculations fast"
sentence_B = "Data science uses NumPy arrays for fast calculations"#AP24110010866

# One unrelated sentence (different topic: pizza)#AP24110010866
sentence_C = "I love eating pizza on weekends"#AP24110010866

# Tokenize each sentence (lowercase + split into words)#AP24110010866
tokens_A = sentence_A.lower().split()   # Break sentence A into lowercase words#AP24110010866
tokens_B = sentence_B.lower().split()   # Break sentence B into lowercase words
tokens_C = sentence_C.lower().split()   # Break sentence C into lowercase words#AP24110010866
#AP24110010866
# Function to turn a sentence's tokens into a bag-of-words vector
def build_bow(tokens, vocab):#AP24110010866
    return np.array([tokens.count(word) for word in vocab])  # Count how many times each vocab word appears#AP24110010866
  #AP24110010866
# Function to calculate cosine similarity between two vectors#AP24110010866
def cosine_similarity(v1, v2):
    dot = np.dot(v1, v2)              # Multiply matching values and add them up#AP24110010866
    mag1 = np.linalg.norm(v1)         # Length (magnitude) of vector 1
    mag2 = np.linalg.norm(v2)         # Length (magnitude) of vector 2
    return dot / (mag1 * mag2)        # Cosine similarity = dot product divided by lengths

# ---- Compare A and B (similar sentences) ----
vocab_AB = sorted(set(tokens_A) | set(tokens_B))   # Combine unique words from A and B
vec_A = build_bow(tokens_A, vocab_AB)              # Vector for sentence A
vec_B = build_bow(tokens_B, vocab_AB)              # Vector for sentence B
similarity_AB = cosine_similarity(vec_A, vec_B)    # Calculate similarity score

# ---- Compare A and C (different sentences) ----
vocab_AC = sorted(set(tokens_A) | set(tokens_C))   # Combine unique words from A and C
vec_A2 = build_bow(tokens_A, vocab_AC)             # Vector for sentence A (new vocab)
vec_C = build_bow(tokens_C, vocab_AC)              # Vector for sentence C
similarity_AC = cosine_similarity(vec_A2, vec_C)   # Calculate similarity score

# Print both results to compare#AP24110010866
print("Similarity A vs B (similar topic):", similarity_AB)
print("Similarity A vs C (different topic):", similarity_AC)
```

Your answer:

Cosine similarity checks how much two sentences overlap in the words they use, not how long the sentences are. If two sentences share many of the same words, their vectors point in almost the same direction, so the similarity score is close to 1. If two sentences share very few or no words, their vectors point in different directions, so the score is close to 0.#AP24110010866

A cosine similarity of exactly 1 means the two vectors point in the exact same direction — the sentences use the same words in the same relative proportions, even if they're not identical in length.

Take any paragraph of your choice (3-4 sentences, write it below as a string). Then:

2.Build its vocabulary and word-count vector (reuse bag_of_words_vector).

```python
import numpy as np  # Import NumPy for array tools and sorting functions

# Step 1: Define the paragraph#AP24110010866
paragraph = "Machine learning is a part of artificial intelligence. Machine learning helps computers learn from data. Data is very important for machine learning models. Good data makes machine learning more accurate."  # Our sample text

# Step 2: Tokenize the paragraph#AP24110010866
clean_text = paragraph.lower().replace(".", "")             # Make everything lowercase and remove periods so words match properly
tokens = clean_text.split()                                 # Split the cleaned text into a list of individual words

print("Tokens:", tokens)            #AP24110010866                        # Print the list of words we got

# Step 3: Build vocabulary and word-count vector
vocab = sorted(set(tokens))                                 # set() keeps only unique words, sorted() arranges them alphabetically
#AP24110010866
def bag_of_words_vector(tokens, vocab):                     # Function that builds a count vector for the vocabulary
    return np.array([tokens.count(word) for word in vocab]) # Counts how many times each vocab word appears in tokens

word_counts = bag_of_words_vector(tokens, vocab)            # Run the function to get the actual counts

print("Vocabulary:", vocab)          # Print the list of unique words
print("Word Counts:", word_counts)   # Print how many times each word appeared
#AP24110010866
# Step 4: Find the top 3 most frequent words
sorted_indices = np.argsort(word_counts)    # Get index positions sorted from lowest count to highest count
top_indices = sorted_indices[::-1][:3]      # Reverse the order to highest-to-lowest, then keep only the first 3

print("\nTop 3 most frequent words:")   # Print a header line
for i in top_indices:                    # Go through each of the top 3 index positions
    print(vocab[i], "-", word_counts[i])  # Print the word at that index and its count
```

Simple Text Similarity Checker (NLP Track)

Build a small program that:#AP24110010866

1.Takes three short sentences of your choice (write them as strings).

2.Tokenizes all three and builds one shared vocabulary across all of them.

3.Converts each sentence into a bag-of-words vector using that shared vocabulary.#AP24110010866

4.Computes cosine similarity between every pair of sentences (3 pairs total: 1-2, 1-3, 2-3).

5.Prints which two sentences are the most similar to each other, based on the highest cosine similarity score.

```python
import numpy as np  # Import NumPy for array math and similarity calculations

# ============================================
# Step 1: Define three sentences#AP24110010866
# ============================================
sentence1 = "I enjoy reading books about space and planets"      # First sentence
sentence2 = "Reading about planets and space is fun for me"      # Second sentence (similar topic to sentence1)
sentence3 = "My favorite food is spicy noodles with vegetables"  # Third sentence (different topic)

# ============================================
# Step 2: Tokenize all three sentences#AP24110010866
# ============================================
tokens1 = sentence1.lower().split()    # Lowercase sentence 1 and split it into a list of words
tokens2 = sentence2.lower().split()    # Lowercase sentence 2 and split it into a list of words
tokens3 = sentence3.lower().split()    # Lowercase sentence 3 and split it into a list of words

# Build one SHARED vocabulary across all three sentences
vocab = sorted(set(tokens1) | set(tokens2) | set(tokens3))   # Combine unique words from all 3 sentences, then sort them

print("Shared Vocabulary:", vocab)   # Print the shared vocabulary list

# ============================================
# Step 3: Convert each sentence to a bag-of-words vector
# ============================================#AP24110010866
def bag_of_words_vector(tokens, vocab):                              # Define function to build a word-count vector
    return np.array([tokens.count(word) for word in vocab])         # For each vocab word, count how many times it's in tokens

vec1 = bag_of_words_vector(tokens1, vocab)   # Build the vector for sentence 1
vec2 = bag_of_words_vector(tokens2, vocab)   # Build the vector for sentence 2
vec3 = bag_of_words_vector(tokens3, vocab)   # Build the vector for sentence 3

print("Vector 1:", vec1)   # Print sentence 1's vector
print("Vector 2:", vec2)   # Print sentence 2's vector
print("Vector 3:", vec3)   # Print sentence 3's vector

# ============================================
# Step 4: Compute cosine similarity for every pair#AP24110010866
# ============================================
def cosine_similarity(v1, v2):                # Define function to calculate cosine similarity between two vectors
    dot_product = np.dot(v1, v2)              # Multiply matching values in both vectors and add them up
    norm1 = np.linalg.norm(v1)                # Calculate the length (magnitude) of vector 1
    norm2 = np.linalg.norm(v2)                # Calculate the length (magnitude) of vector 2
    return dot_product / (norm1 * norm2)      # Divide dot product by both lengths to get cosine similarity

sim_1_2 = cosine_similarity(vec1, vec2)   # Compare sentence 1 and sentence 2
sim_1_3 = cosine_similarity(vec1, vec3)   # Compare sentence 1 and sentence 3
sim_2_3 = cosine_similarity(vec2, vec3)   # Compare sentence 2 and sentence 3

print("\nSimilarity (1 vs 2):", sim_1_2)   # Print similarity score for pair 1-2
print("Similarity (1 vs 3):", sim_1_3)     # Print similarity score for pair 1-3#AP24110010866
print("Similarity (2 vs 3):", sim_2_3)     # Print similarity score for pair 2-3

# ============================================
# Step 5: Find and print the most similar pair
# ============================================
# Store all pairs with their similarity scores in a dictionary
similarities = {
    "Sentence 1 & 2": sim_1_2,   # Store pair 1-2's score with a label#AP24110010866
    "Sentence 1 & 3": sim_1_3,   # Store pair 1-3's score with a label
    "Sentence 2 & 3": sim_2_3    # Store pair 2-3's score with a label
}

most_similar_pair = max(similarities, key=similarities.get)   # Find the dictionary key with the highest score

print(f"\nMost similar pair: {most_similar_pair} (score: {similarities[most_similar_pair]:.4f})")  # Print the winning pair and its score#AP24110010866
```

Reflection Questions

1. Why do we need to convert text into numbers before a machine learning model can use it?

Machine learning models are built on mathematical operations like multiplication, addition, and comparisons — they can't directly understand words or sentences as text. Converting words into numbers (like a bag-of-words vector) gives the model something it can actually calculate with, such as measuring distances or similarities between pieces of text.#AP24110010866

2. What information does a bag-of-words vector lose about the original sentence?

It loses word order and grammar completely — "dog bites man" and "man bites dog" would produce nearly identical vectors even though they mean very different things. It also loses context and meaning between words, punctuation, and any sense of how words relate to each other in a sentence — it only cares about which words appear and how many times, not how they're arranged.#AP24110010866

3. Why might cosine similarity be a better similarity measure than simply comparing raw word-count differences?

Cosine similarity focuses on the direction of the vectors (the pattern of word usage) rather than their raw size, so it isn't thrown off by sentence length. A short sentence and a long sentence discussing the same topic can still score highly similar, because cosine similarity normalizes for magnitude. Comparing raw word-count differences directly would unfairly penalize longer sentences just for having bigger numbers, even if they're about the same topic.

4. Name one real-world application that likely uses something like this text-vectorization + similarity approach.
#AP24110010866
Search engines use this kind of approach: when you type a search query, the engine converts your query into a vector and compares it against vectors representing web pages or documents, then returns the results whose vectors are most similar (highest cosine similarity) to what you typed. This is a simplified version of the real ranking systems used, but the core idea of vectorizing text and measuring similarity is foundational to how search relevance works.
